The aim of this notebook is to accurately identify the species of Iris flower using a Support Vector Machine model, I compare the results of using the exact same training & test sets with only the kernel used being Linear or Radial Basis Function (RBF)

Dataset used: https://www.kaggle.com/datasets/vikrishnan/iris-dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.svm import SVC
```

### Dropped ID column because its not needed

```python
df = pd.read_csv('Iris.csv')
df = df.drop('Id', axis='columns')
```

```python
xAxis = df[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']]
yAxis = df['Species']
```

```python
scaler = StandardScaler()
xScaled = scaler.fit_transform(xAxis) #Balances the influcence of each feature
```

```python
xTrain, xTest, yTrain, yTest = train_test_split(
    xScaled, yAxis, test_size=0.31, random_state=10 #Tried several different test size values, found 31% to give the highest model score
)
```

```python
model = SVC(kernel='rbf') #Switched kernel from linear to rbf, model score now 100%, up from 97%
model.fit(xTrain, yTrain)
```

```python
xVisualised = xScaled[:, :2]
modelVisualised = SVC(kernel='rbf').fit(xVisualised, yAxis)
print(model.score(xTest, yTest))
```

### Confusion Matrix shows 100% accuracy

```python
xPlot = xScaled[:, :2] # Using SepalLengthCm & SepalWidthCm
yPlot = yAxis

yPredict = model.predict(xTest)
confusionMatrix = confusion_matrix(yTest, yPredict)
print("Confusion Matrix:\n", confusionMatrix)
```

```python
encoder = LabelEncoder()
yPlotEncoded = encoder.fit_transform(yPlot)

modelPlot = SVC(kernel='rbf', gamma='scale')
modelPlot.fit(xPlot, yPlotEncoded)


x_min, x_max = xPlot[:, 0].min() - 1, xPlot[:, 0].max() + 1
y_min, y_max = xPlot[:, 1].min() - 1, xPlot[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 500),
                     np.linspace(y_min, y_max, 500))

Z = modelPlot.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
```

```python
plt.contourf(xx, yy, Z, alpha=0.2, cmap=plt.cm.coolwarm)


for idx, species in enumerate(np.unique(yPlot)):
    plt.scatter(xPlot[yPlot == species, 0],
                xPlot[yPlot == species, 1],
                label=species,
                edgecolor='k')

plt.xlabel("Sepal Length (scaled)")
plt.ylabel("Sepal Width (scaled)")
plt.title("SVM Decision Boundary (RBF Kernel)")
plt.legend()
plt.show()
```

```python
model = SVC(kernel='linear') #Switched kernel from linear to rbf, model score now 100%, up from 97%
model.fit(xTrain, yTrain)
```
